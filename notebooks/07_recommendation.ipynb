```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('online_retail_sales_cleaned.csv')
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
```

```python
df.shape
```

```python
df.head(10)
```

```python
popular_products = df.groupby('Description')['Quantity'].sum().sort_values(ascending=False)
```

```python
top_10_popular = popular_products.head(10)
```

```python
top_10_popular.head(10)
```

```python
top_100_products = df.groupby('Description')['Quantity'].sum().sort_values(ascending=False).head(100).index
df_sample = df[df['Description'].isin(top_100_products)]
```

```python
top_100_products
```

```python
df_sample.shape
```

```python
customer_product_matrix = df_sample.pivot_table(
    index='Customer ID',
    columns='Description',
    values='Quantity',
    aggfunc='sum',
    fill_value=0
)
```

```python
customer_product_matrix.shape
```

```python
customer_product_matrix.head()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
product_similarity = cosine_similarity(customer_product_matrix.T)
```

```python
product_similarity.shape
```

```python
product_similarity_df = pd.DataFrame(
    product_similarity,
    index=customer_product_matrix.columns,
    columns=customer_product_matrix.columns
)
```

```python
product_similarity_df.shape
```

```python
product_similarity_df.head()
```

```python
def get_similar_products(product_name, top_n=10):
    similar_scores = product_similarity_df[product_name].sort_values(ascending=False)
    similar_scores = similar_scores.drop(product_name)
    return similar_scores.head(top_n)
```

```python
get_similar_products('WHITE HANGING HEART T-LIGHT HOLDER')
```

```python
product_similarity_df.to_csv('product_similarity_matrix.csv')
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('online_retail_II.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=['Customer ID'])
```

```python
df['Description'] = df['Description'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
cancelled = df[df['Invoice'].astype(str).str.startswith('C')]
print('Cancelled/Returned invoices rows:', cancelled.shape[0])
```

```python
df_sales = df[df['Quantity'] > 0].copy()
print('Sales rows:', df_sales.shape[0])
```

```python
df_returns = df[df['Quantity'] < 0].copy()
print('Returns/Cancelled rows:', df_returns.shape[0])
```

```python
zero_price = df_sales[df_sales['Price'] == 0]
print('Zero price rows:', zero_price.shape[0])
```

```python
df_sales = df_sales[df_sales['Price'] > 0].copy()
print('After removing zero-price rows:', df_sales.shape[0])
```

```python
df_sales['InvoiceDate'] = pd.to_datetime(df_sales['InvoiceDate'])
print('Setting DataTypes Of InvoiceDate', df_sales['InvoiceDate'])
```

```python
df_sales['Customer ID'] = df_sales['Customer ID'].astype(int).astype(str)
df_sales.head(10)
```

```python
df.info()
```

```python
df_sales['TotalPrice'] = df_sales['Quantity'] * df_sales['Price']
df_sales[['Quantity', 'Price', 'TotalPrice']].head(10)
```

```python
df_sales.to_csv('online_retail_sales_cleaned.csv', index=False)
df_returns.to_csv('online_retail_returns.csv', index=False)
```

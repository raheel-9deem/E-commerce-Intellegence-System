```python
import pandas as pd
```

```python
df = pd.read_csv('rfm_segments.csv')
```

```python
df.head()
```

```python
df['AvgOrderValue'] = df['Monetary'] / df['Frequency']
```

```python
df.head(10)
```

```python
df.columns.tolist()
```

```python
df['Churned'] = (df['Recency'] > 90).astype(int)
```

```python
df.columns.tolist()
```

```python
df.head(10)
```

```python
df.to_csv('rfm_features_full.csv', index=False)
```

```python
churn_rate = df['Churned'].mean()
```

```python
print(churn_rate)
```

```python
df.head()
```

```python
estimated_lifespan_months = 1 / churn_rate
```

```python
print(estimated_lifespan_months)
```

```python
df['CLV'] = df['AvgOrderValue'] * df['Frequency'] * estimated_lifespan_months
df[['Customer ID', 'AvgOrderValue', 'Frequency', 'CLV']].sort_values('CLV', ascending=False).head(10)
```

```python
df.to_csv('clv_results.csv', index=False)
```
